```python
from pathlib import Path
import pandas as pd
import numpy as np
import json

BASE_DIR = Path(
    r"D:\DATA SCIENCE COURSE\ML\ML_Projects\WasteWise-AI"
)

RAW_DIR = BASE_DIR / "datasets" / "raw"
OUT_DIR = BASE_DIR / "datasets" / "processed"
REPORT_DIR = BASE_DIR / "datasets" / "reports"

OUT_DIR.mkdir(parents=True, exist_ok=True)
REPORT_DIR.mkdir(parents=True, exist_ok=True)

print("Raw folder exists:", RAW_DIR.exists())
print("Output folder:", OUT_DIR)
print("Reports folder:", REPORT_DIR)
```

```python
file_path = RAW_DIR / "inventory_batches.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)
```

```python
display(df.head())

print("Column names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nExact duplicate rows:", df.duplicated().sum())
```

```python
print("Unique batch IDs:", df["batch_id"].nunique())
print("Unique product IDs:", df["product_id"].nunique())

print("\nExpiry risk label counts:")
print(df["expiry_risk_label"].value_counts(dropna=False))

print("\nDays to expiry range:")
print(df["days_to_expiry"].min())
print(df["days_to_expiry"].max())
```

```python
text_columns = [
    "batch_id",
    "product_id"
]

for column in text_columns:
    df[column] = df[column].astype("string").str.strip()

display(df.head())
```

```python
print(df.columns.tolist())
```

```python
date_columns = [
    "received_at",
    "snapshot_date",
    "expiry_date"
]

for column in date_columns:
    df[column] = pd.to_datetime(
        df[column],
        errors="coerce"
    )

print(df[date_columns].dtypes)
display(df[date_columns].head())
```

```python
numeric_columns = [
    "initial_quantity",
    "quantity_remaining",
    "days_to_expiry",
    "avg_daily_sales_7d",
    "expected_demand_before_expiry",
    "unit_price",
    "unit_cost",
    "discount_pct",
    "expiry_risk_label",
    "waste_units_actual_simulated"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )

print(df[numeric_columns].dtypes)
```

```python
numeric_columns = [
    "initial_quantity",
    "quantity_remaining",
    "days_to_expiry",
    "avg_daily_sales_7d",
    "expected_demand_before_expiry",
    "unit_price",
    "unit_cost",
    "discount_pct",
    "expiry_risk_label",
    "waste_units_actual_simulated"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )

print(df[numeric_columns].dtypes)
```

```python
print("Missing values after conversion:")
print(df.isnull().sum())

print("\nDuplicate batch IDs:")

print(df["batch_id"].duplicated().sum())
```

```python
invalid_mask = (
    df["batch_id"].isna()
    | df["batch_id"].eq("")
    | df["product_id"].isna()
    | df["product_id"].eq("")
    | df["category"].isna()
    | df["category"].eq("")
    | df[date_columns].isnull().any(axis=1)
    | df[numeric_columns].isnull().any(axis=1)
    | df["initial_quantity"].lt(0)
    | df["quantity_remaining"].lt(0)
    | df["unit_price"].lt(0)
    | df["unit_cost"].lt(0)
    | df["waste_units_actual_simulated"].lt(0)
    | df["avg_daily_sales_7d"].lt(0)
    | df["expected_demand_before_expiry"].lt(0)
    | ~df["expiry_risk_label"].isin([0, 1])
    | df["quantity_remaining"].gt(df["initial_quantity"])
    | df["waste_units_actual_simulated"].gt(
        df["quantity_remaining"]
    )
    | (df["received_at"] > df["snapshot_date"])
)

print("Invalid rows identified:", invalid_mask.sum())
```

```python
calculated_days = (
    df["expiry_date"] - df["snapshot_date"]
).dt.days

expiry_mismatch_mask = (
    df["days_to_expiry"] != calculated_days
)

print(
    "Expiry calculation mismatches:",
    expiry_mismatch_mask.sum()
)
```

```python
duplicate_batch_mask = df["batch_id"].duplicated(
    keep=False
)

duplicate_batches = df[duplicate_batch_mask].copy()

print("Rows with duplicate batch IDs:", len(duplicate_batches))

display(duplicate_batches.head())
```

```python
invalid_mask = invalid_mask | expiry_mismatch_mask

rejected_df = df[invalid_mask].copy()
clean_df = df[~invalid_mask].copy()

print("Original rows:", len(df))
print("Rejected rows:", len(rejected_df))
print("Rows remaining:", len(clean_df))
```

```python

before = len(clean_df)

clean_df = clean_df.drop_duplicates().copy()

after = len(clean_df)

print("Exact duplicate rows removed:", before - after)

print("Rows after removing exact duplicates:", after)
```

```python
duplicate_ids = clean_df["batch_id"].duplicated(
    keep=False
)

duplicate_id_rows = clean_df[duplicate_ids].copy()

clean_df = clean_df[~duplicate_ids].copy()

print("Duplicate-ID rows sent for review:", len(duplicate_id_rows))
print("Final clean rows:", len(clean_df))
```

```python
clean_df = clean_df.sort_values(
    ["product_id", "snapshot_date", "expiry_date"]
).reset_index(drop=True)

clean_path = OUT_DIR / "inventory_batches_clean.csv"
rejected_path = REPORT_DIR / "inventory_batches_rejected.csv"
duplicate_path = REPORT_DIR / "inventory_batches_duplicate_ids.csv"
report_path = REPORT_DIR / "inventory_batches_report.json"

clean_df.to_csv(clean_path, index=False)
rejected_df.to_csv(rejected_path, index=False)
duplicate_id_rows.to_csv(duplicate_path, index=False)

report = {
    "original_rows": int(len(df)),
    "clean_rows": int(len(clean_df)),
    "rejected_rows": int(len(rejected_df)),
    "duplicate_batch_id_rows": int(len(duplicate_id_rows)),
    "clean_columns": clean_df.columns.tolist()
}

with open(report_path, "w", encoding="utf-8") as file:
    json.dump(report, file, indent=4)

print("Cleaned dataset saved to:", clean_path)
print("Rejected rows saved to:", rejected_path)
print("Duplicate batch IDs saved to:", duplicate_path)

print("Cleaning report saved to:", report_path)
```
