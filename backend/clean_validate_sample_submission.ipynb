```python
from pathlib import Path
import pandas as pd
import json

BASE_DIR = Path(
    r"D:\DATA SCIENCE COURSE\ML\ML_Projects\WasteWise-AI"
)

RAW_DIR = BASE_DIR / "datasets" / "raw"
OUT_DIR = BASE_DIR / "datasets" / "processed"
REPORT_DIR = BASE_DIR / "datasets" / "reports"

OUT_DIR.mkdir(parents=True, exist_ok=True)
REPORT_DIR.mkdir(parents=True, exist_ok=True)

print("Raw folder exists:", RAW_DIR.exists())
```

```python
file_path = RAW_DIR / "sample_submission.csv"

df = pd.read_csv(file_path)

print("File loaded successfully!")
print("Shape:", df.shape)
```

```python
display(df.head())

print("Column names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nExact duplicate rows:", df.duplicated().sum())
```

```python
if "id" in df.columns:
    print("Unique IDs:", df["id"].nunique())
    print("Missing IDs:", df["id"].isna().sum())
    print("Duplicate IDs:", df["id"].duplicated().sum())
else:
    print("No 'id' column found. Check the column names above.")
```

```python
if "id" in df.columns:
    forecast_columns = [
        column for column in df.columns
        if column != "id"
    ]

    print("Number of forecast columns:", len(forecast_columns))
    print("Forecast columns:")
    print(forecast_columns)

    print("\nMissing values in forecast columns:",
          df[forecast_columns].isnull().sum().sum())
```

```python
issues = []

if "id" not in df.columns:
    issues.append("Missing required id column")
else:
    if df["id"].isna().any():
        issues.append("Missing IDs found")

    if df["id"].duplicated().any():
        issues.append("Duplicate IDs found")

if df.empty:
    issues.append("The file contains no rows")

print("Validation issues:")

if issues:
    for issue in issues:
        print("-", issue)
else:
    print("No structural issues found.")
```

```python
output_path = OUT_DIR / "sample_submission_validated.csv"
report_path = REPORT_DIR / "sample_submission_report.json"

df.to_csv(output_path, index=False)

report = {
    "rows": int(len(df)),
    "columns": int(len(df.columns)),
    "duplicate_rows": int(df.duplicated().sum()),
    "missing_values": int(df.isnull().sum().sum()),
    "validation_issues": issues,
    "column_names": df.columns.tolist()
}

with open(report_path, "w", encoding="utf-8") as file:
    json.dump(report, file, indent=4)

print("Validated template saved to:", output_path)
print("Report saved to:", report_path)
```

```python
validated_df = pd.read_csv(output_path)

print("Final shape:", validated_df.shape)
print("Columns preserved:", validated_df.columns.equals(df.columns))
display(validated_df.head())
```
