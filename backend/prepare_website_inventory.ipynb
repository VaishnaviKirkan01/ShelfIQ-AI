```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path(
    r"D:\DATA SCIENCE COURSE\ML\ML_Projects\WasteWise-AI"
)

PROCESSED_DIR = BASE_DIR / "datasets" / "processed"
REPORT_DIR = BASE_DIR / "datasets" / "reports"

REPORT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
products_df = pd.read_csv(
    PROCESSED_DIR / "products_clean.csv"
)

batches_df = pd.read_csv(
    PROCESSED_DIR / "inventory_batches_clean.csv"
)

print("Products:", products_df.shape)
print("Batches:", batches_df.shape)

print("\nProduct columns:")
print(products_df.columns.tolist())

print("\nBatch columns:")
print(batches_df.columns.tolist())
```

```python
assert "product_id" in products_df.columns
assert "product_id" in batches_df.columns

assert products_df["product_id"].is_unique, (
    "Products contain duplicate product IDs."
)

print("Product IDs are unique.")
```

```python
inventory_df = batches_df.merge(
    products_df,
    on="product_id",
    how="left",
    validate="many_to_one",
    suffixes=("_batch", "_product")
)

print("Joined inventory shape:", inventory_df.shape)
display(inventory_df.head())
```

```python
print("Missing values after join:")
print(inventory_df.isnull().sum())

print(
    "\nRows with no matching product:",
    inventory_df["product_id"].isna().sum()
)

print("Total inventory batches:", len(batches_df))
print("Joined inventory rows:", len(inventory_df))
```

```python
output_path = (
    PROCESSED_DIR / "website_inventory.csv"
)

inventory_df.to_csv(output_path, index=False)

print("Website inventory saved to:")
print(output_path)
```

```python
website_inventory = pd.read_csv(output_path)

print("Saved dataset shape:", website_inventory.shape)
print("Columns:", website_inventory.columns.tolist())

display(website_inventory.head())
```
